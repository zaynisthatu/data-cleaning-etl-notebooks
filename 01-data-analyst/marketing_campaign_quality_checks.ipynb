```python
# Import the pandas library for data analysis
import pandas as pd

# Load the dataset from the CSV file
# Make sure to upload 'marketing_campaign_dataset.csv' to your Colab session
file_path = 'marketing_campaign_dataset.csv'
df = pd.read_csv(file_path)

# --- 1. Check for Incorrect Data Types ---
print("## 1. Data Type Inspection ##")
print("Notice that 'Date', 'Acquisition_Cost', and 'Duration' are incorrectly typed as 'object' (text).")
print("-" * 50)
df.info()
print("\n" * 2)


# --- 2. View Formatting Issues ---
print("## 2. Formatting Issues in Columns ##")
print("You can see the '$' and commas in 'Acquisition_Cost' and 'days' in 'Duration'.")
print("-" * 50)
print(df[['Acquisition_Cost', 'Duration']].head())
print("\n" * 2)


# --- 3. Check for Inconsistent Column Names ---
print("## 3. Column Naming Convention ##")
print("The column names use underscores and mixed capitalization.")
print("-" * 50)
print(df.columns.tolist())
print("\n" * 2)


# --- 4. Check for Null or Missing Values ---
print("## 4. Null Value Check ##")
print("This confirms that there are no missing values in the dataset.")
print("-" * 50)
print(df.isnull().sum())
print("\n" * 2)


# --- 5. Check for Duplicate Rows ---
print("## 5. Duplicate Row Check ##")
print("This confirms there are no duplicate rows in the dataset.")
print("-" * 50)
duplicate_rows = df.duplicated().sum()
print(f"Found {duplicate_rows} duplicate rows.")
```

```python
import pandas as pd

# Load the original dataset if you haven't already
file_path = 'marketing_campaign_dataset.csv'
df = pd.read_csv(file_path)

# Create a copy to keep the original data unchanged
df_cleaned = df.copy()

# --- 1. Standardize Column Names ---
# Converts all column names to lowercase and removes underscores
print("## 1. Standardizing Column Names ##")
original_columns = df_cleaned.columns.tolist()
df_cleaned.columns = df_cleaned.columns.str.lower().str.replace('_', '')
print("Original column names:", original_columns)
print("Cleaned column names:", df_cleaned.columns.tolist())
print("-" * 50, "\n")


# --- 2. Clean and Convert 'acquisitioncost' Column ---
print("## 2. Cleaning 'acquisitioncost' ##")
# Remove '$' and ',' then convert to a numeric (float) type
df_cleaned['acquisitioncost'] = df_cleaned['acquisitioncost'].replace({'\$': '', ',': ''}, regex=True).astype(float)
print("First 5 cleaned 'acquisitioncost' values:")
print(df_cleaned['acquisitioncost'].head())
print("-" * 50, "\n")


# --- 3. Clean and Convert 'duration' Column ---
print("## 3. Cleaning 'duration' ##")
# Remove ' days' and convert to an integer type
df_cleaned['duration'] = df_cleaned['duration'].str.replace(' days', '').astype(int)
print("First 5 cleaned 'duration' values:")
print(df_cleaned['duration'].head())
print("-" * 50, "\n")


# --- 4. Convert 'date' Column ---
print("## 4. Converting 'date' Column ##")
# Convert the 'date' column to datetime objects
df_cleaned['date'] = pd.to_datetime(df_cleaned['date'])
print("Data type of 'date' column after conversion:")
print(df_cleaned['date'].dtype)
print("-" * 50, "\n")


# --- Verification ---
print("## Verification: Final Data Types ##")
print("All columns now have the correct data types and formatting.")
print("-" * 50)
df_cleaned.info()


# --- Display Cleaned Data ---
print("\n\n## First 5 Rows of the Cleaned DataFrame ##")
print(df_cleaned.head())


# --- Optional: Save the Cleaned Data to a New File ---
# You can uncomment the line below to save your cleaned data to a new CSV file.
# df_cleaned.to_csv('marketing_campaign_cleaned.csv', index=False)
# print("\nCleaned data has been saved to 'marketing_campaign_cleaned.csv'")
```
